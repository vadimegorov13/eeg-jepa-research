# Liu2024 Stroke MI CSP/FBCSP Baselines

This notebook is the Liu2024 acute-stroke motor-imagery version of the existing CSP/FBCSP baseline workflow.

The purpose is to answer a narrow sanity question before running S-JEPA:

> Are the Liu2024 stroke-patient left/right MI windows learnable with classical MI decoding?

It keeps the same structure as the previous baseline notebook:

- Dataset: `Liu2024` through MOABB / Braindecode
- Labels: `left_hand` vs `right_hand`
- Average reference before preprocessing
- EEG only, resample to 128 Hz, bandpass 0.5--40 Hz
- Explicit 537-sample windowing with CSP-specific non-overlap event filtering
- Same post-window µV scaling
- Same within-subject 5-fold stratified CV with seed 12
- Same artifact style: `config.json`, `cv_results.json`, `subject_metrics.json`, `global_metrics.json`, `collapse_diagnostics.json`, `run_metadata.json`, and `run.log`

```python
LOG_PATH = ARTIFACT_DIR / "run.log"
_LOG_FILE_HANDLE = open(LOG_PATH, "a", buffering=1)

def _timestamped_print(*args, **kwargs):
    sep = kwargs.pop("sep", " ")
    end = kwargs.pop("end", "\n")
    flush = kwargs.pop("flush", False)
    file = kwargs.pop("file", None)

    message = sep.join(str(arg) for arg in args)

    leading_newlines = len(message) - len(message.lstrip("\n"))
    message_body = message[leading_newlines:]

    def _write_target(text):
        if file is None:
            sys.__stdout__.write(text)
            if flush:
                sys.__stdout__.flush()
        else:
            file.write(text)
            if flush and hasattr(file, "flush"):
                file.flush()

    if leading_newlines > 0:
        blanks = "\n" * leading_newlines
        _write_target(blanks)
        _LOG_FILE_HANDLE.write(blanks)

    if message_body:
        ts = datetime.now().strftime("%Y-%m-%d %H:%M:%S")
        stamped = f"[{ts}] {message_body}"
        _write_target(stamped + end)
        _LOG_FILE_HANDLE.write(stamped + end)
    else:
        _write_target(end)
        _LOG_FILE_HANDLE.write(end)

    if flush:
        _LOG_FILE_HANDLE.flush()

builtins.print = _timestamped_print
print(f"Logging to: {LOG_PATH}")
```

## 3.4. Windowing

```python
def summarize_annotation_counts(dataset):
    counts = {}
    for ds in dataset.datasets:
        descriptions = np.asarray(ds.raw.annotations.description).astype(str)
        for desc in descriptions:
            counts[desc] = counts.get(desc, 0) + 1
    return dict(sorted(counts.items()))


def filter_annotations_by_description(dataset, descriptions_to_keep):
    keep_set = set(str(d) for d in descriptions_to_keep)
    total_before = 0
    total_after = 0

    for ds in dataset.datasets:
        annotations = ds.raw.annotations
        if len(annotations) == 0:
            continue

        descriptions = np.asarray(annotations.description).astype(str)
        mask = np.isin(descriptions, list(keep_set))
        total_before += len(annotations)
        total_after += int(np.sum(mask))

        filtered_annotations = mne.Annotations(
            onset=np.asarray(annotations.onset)[mask],
            duration=np.asarray(annotations.duration)[mask],
            description=descriptions[mask].tolist(),
            orig_time=annotations.orig_time,
        )
        ds.raw.set_annotations(filtered_annotations)

    return total_before, total_after


def collect_selected_annotation_timing(dataset):
    """Collect timing rows after annotation labels have been filtered."""
    rows = []
    for ds_idx, ds in enumerate(dataset.datasets):
        annotations = ds.raw.annotations
        if len(annotations) == 0:
            continue

        onsets = np.asarray(annotations.onset, dtype=float)
        durations = np.asarray(annotations.duration, dtype=float)
        descriptions = np.asarray(annotations.description).astype(str)

        order = np.argsort(onsets)
        onsets = onsets[order]
        durations = durations[order]
        descriptions = descriptions[order]

        subject_id = str(ds.description.get("subject", ds_idx))
        for event_idx, (onset, duration, desc) in enumerate(zip(onsets, durations, descriptions)):
            rows.append(
                {
                    "recording_index": ds_idx,
                    "subject_id": subject_id,
                    "event_index": event_idx,
                    "onset_s": float(onset),
                    "duration_s": float(duration),
                    "description": str(desc),
                }
            )
    return rows


def print_annotation_timing_diagnostic(dataset, target_window_samples, sfreq, max_examples=10):
    rows = collect_selected_annotation_timing(dataset)
    if len(rows) == 0:
        print("Annotation timing diagnostic: no selected annotations found.")
        return None

    durations = np.asarray([r["duration_s"] for r in rows], dtype=float)
    effective_duration_s = target_window_samples / float(sfreq)

    print("Annotation timing diagnostic after label filtering:")
    print(f"  Selected events:             {len(rows)}")
    print(f"  Duration min/median/max:     {np.min(durations):.6f} / {np.median(durations):.6f} / {np.max(durations):.6f} s")
    print(f"  Requested analysis samples:  {target_window_samples}")
    print(f"  Effective analysis duration: {effective_duration_s:.6f} s")

    overlap_examples = []
    all_gaps = []
    by_recording = {}
    for r in rows:
        by_recording.setdefault(r["recording_index"], []).append(r)

    for ds_idx, rec_rows in by_recording.items():
        rec_rows = sorted(rec_rows, key=lambda x: x["onset_s"])
        for left, right in zip(rec_rows[:-1], rec_rows[1:]):
            gap = float(right["onset_s"] - left["onset_s"])
            all_gaps.append(gap)
            if gap < effective_duration_s:
                overlap_examples.append((ds_idx, left, right, gap))

    if all_gaps:
        gaps = np.asarray(all_gaps, dtype=float)
        print(f"  Onset gap min/median/max:    {np.min(gaps):.6f} / {np.median(gaps):.6f} / {np.max(gaps):.6f} s")
    else:
        print("  Onset gap summary:           only one selected event per recording")

    print(f"  Potential {effective_duration_s:.6f}s overlaps: {len(overlap_examples)}")
    for ds_idx, left, right, gap in overlap_examples[:max_examples]:
        print(
            "    overlap example "
            f"recording={ds_idx}, subject={left['subject_id']}, "
            f"{left['description']}@{left['onset_s']:.3f}s -> "
            f"{right['description']}@{right['onset_s']:.3f}s, gap={gap:.3f}s"
        )

    return {
        "n_events": len(rows),
        "median_duration_s": float(np.median(durations)),
        "effective_duration_s": float(effective_duration_s),
        "n_potential_overlaps": int(len(overlap_examples)),
    }


def annotation_class_counts(dataset, labels_to_keep):
    labels = list(labels_to_keep)
    counts = {str(label): 0 for label in labels}
    for ds in dataset.datasets:
        descriptions = np.asarray(ds.raw.annotations.description).astype(str)
        for desc in descriptions:
            if str(desc) in counts:
                counts[str(desc)] += 1
    return counts


def annotation_duration_sample_summary(dataset):
    """Return min/median/max annotation durations after converting durations to samples."""
    sample_counts = []
    for ds in dataset.datasets:
        sfreq = float(ds.raw.info["sfreq"])
        durations = np.asarray(ds.raw.annotations.duration, dtype=float)
        for duration_s in durations:
            # Use floor because some nominal 4.0s Liu annotations become 511 samples after conversion.
            sample_counts.append(int(math.floor(float(duration_s) * sfreq)))

    if len(sample_counts) == 0:
        raise RuntimeError("No annotation durations available for sample summary.")

    arr = np.asarray(sample_counts, dtype=int)
    return {
        "min": int(np.min(arr)),
        "median": int(np.median(arr)),
        "max": int(np.max(arr)),
    }


def compute_required_nonoverlap_gap_samples(dataset, target_window_samples):
    """
    Compute the onset-gap rule used before Braindecode windowing.

    There are two useful modes for Liu2024:
      - major_overlap_only: drop severe ~2s overlaps, but keep normal ~4s-spaced events.
        This preserves many more trials while still reducing the most obvious label contamination.
        Because 537 samples is longer than a nominal 4s annotation, Braindecode may still warn
        about the small ~25-sample tail overlap. That warning is expected in this mode.
      - strict_nonoverlap: require no overlap after the trial_stop_offset extension. This is the
        cleanest supervised CSP interpretation, but it often leaves only about 12 windows/subject.
    """
    summary = annotation_duration_sample_summary(dataset)
    mode = str(CONFIG.get("overlap_filter_mode", "major_overlap_only"))

    if mode == "strict_nonoverlap":
        provisional_stop_offset = int(target_window_samples - summary["min"])
        required_gap_samples = int(summary["max"] + max(provisional_stop_offset, 0))
        reason = (
            "strict_nonoverlap: require enough onset distance for the full adjusted "
            "Braindecode trial interval. This may discard many Liu2024 events."
        )
    elif mode == "major_overlap_only":
        allowed_overlap = CONFIG.get("max_allowed_overlap_with_next_event_samples", None)
        if allowed_overlap is None:
            # With 537-sample target and nominal 512-sample Liu annotations, this becomes 25 samples.
            # Therefore the required onset gap becomes 512 samples: it drops ~2s gaps but keeps ~4s gaps.
            allowed_overlap = max(int(target_window_samples - summary["median"]), 0)
        allowed_overlap = int(max(0, allowed_overlap))
        required_gap_samples = int(max(1, target_window_samples - allowed_overlap))
        provisional_stop_offset = int(target_window_samples - summary["min"])
        reason = (
            "major_overlap_only: drop severe overlaps while allowing the small tail overlap "
            "introduced by using a 537-sample window on nominal 4s annotations."
        )
    else:
        raise ValueError(
            "CONFIG['overlap_filter_mode'] must be 'major_overlap_only' or 'strict_nonoverlap'."
        )

    print("CSP overlap-filter gap configuration:")
    print(f"  Mode:                          {mode}")
    print(f"  Duration samples min/median/max: {summary['min']} / {summary['median']} / {summary['max']}")
    print(f"  Target window samples:           {target_window_samples}")
    print(f"  Provisional stop offset:         {provisional_stop_offset}")
    if mode == "major_overlap_only":
        print(f"  Allowed small overlap samples:   {allowed_overlap}")
    print(f"  Required onset gap samples:      {required_gap_samples}")
    print(f"  Reason: {reason}")

    return required_gap_samples


def _make_annotations_from_indices(annotations, keep_indices):
    keep_indices = np.asarray(sorted(set(int(i) for i in keep_indices)), dtype=int)
    if len(keep_indices) == 0:
        return mne.Annotations(onset=[], duration=[], description=[], orig_time=annotations.orig_time)

    descriptions = np.asarray(annotations.description).astype(str)
    return mne.Annotations(
        onset=np.asarray(annotations.onset, dtype=float)[keep_indices],
        duration=np.asarray(annotations.duration, dtype=float)[keep_indices],
        description=descriptions[keep_indices].tolist(),
        orig_time=annotations.orig_time,
    )


def _greedy_nonoverlap_indices(onset_samples, start_index, min_gap_samples):
    """Greedy non-overlap pass starting at one event index."""
    keep = []
    last_onset_sample = None

    for i in range(start_index, len(onset_samples)):
        onset_sample = int(onset_samples[i])
        if last_onset_sample is None or (onset_sample - last_onset_sample) >= int(min_gap_samples):
            keep.append(i)
            last_onset_sample = onset_sample

    return keep


def _score_nonoverlap_candidate(candidate_indices, descriptions, labels_to_keep):
    """
    Score a candidate subset for CSP/FBCSP.

    Priority:
      1. maximize the smallest class count
      2. minimize class imbalance
      3. maximize total kept events
    """
    labels = [str(label) for label in labels_to_keep]
    candidate_desc = [str(descriptions[i]) for i in candidate_indices]
    counts = [candidate_desc.count(label) for label in labels]

    min_count = min(counts) if counts else 0
    imbalance = max(counts) - min(counts) if counts else 0
    total = len(candidate_indices)
    return (min_count, -imbalance, total)


def choose_balanced_nonoverlapping_indices(annotations, min_gap_samples, sfreq, labels_to_keep):
    """
    Select a non-overlapping subset of annotations while trying to preserve class balance.

    This is intentionally conservative for CSP/FBCSP. It keeps a 537-sample analysis window,
    but removes events whose onset would cause raw-sample overlap with the previous kept event.
    A simple greedy pass can accidentally keep mostly one class on Liu2024, so we try each
    possible starting event and select the candidate with the best class balance.
    """
    if len(annotations) == 0:
        return []

    onsets = np.asarray(annotations.onset, dtype=float)
    descriptions = np.asarray(annotations.description).astype(str)
    order = np.argsort(onsets)

    sorted_onsets = onsets[order]
    sorted_descriptions = descriptions[order]
    # Use sample-space comparison so it matches the 537-sample requirement.
    onset_samples = np.asarray(np.round(sorted_onsets * float(sfreq)), dtype=int)

    best_sorted_indices = []
    best_score = None

    for start_index in range(len(order)):
        candidate = _greedy_nonoverlap_indices(
            onset_samples=onset_samples,
            start_index=start_index,
            min_gap_samples=min_gap_samples,
        )
        score = _score_nonoverlap_candidate(
            candidate_indices=candidate,
            descriptions=sorted_descriptions,
            labels_to_keep=labels_to_keep,
        )
        if best_score is None or score > best_score:
            best_score = score
            best_sorted_indices = candidate

    # Convert back from sorted positions to original annotation indices.
    return [int(order[i]) for i in best_sorted_indices]


def drop_overlapping_annotations_for_csp(dataset, target_window_samples, min_gap_samples, labels_to_keep):
    """
    Remove or reduce overlapping labeled events before Braindecode windowing.

    In strict_nonoverlap mode this lets us keep `on_overlapping_events='raise'`.
    In major_overlap_only mode this reduces severe overlaps while allowing the small tail
    overlap needed for a 537-sample window.
    """
    if not CONFIG.get("drop_overlapping_events_for_csp", False):
        print("CSP overlap filtering: disabled by CONFIG.")
        return dataset

    total_before = 0
    total_after = 0
    per_recording = []

    for ds_idx, ds in enumerate(dataset.datasets):
        annotations = ds.raw.annotations
        n_before = len(annotations)
        total_before += n_before

        if n_before == 0:
            per_recording.append((ds_idx, str(ds.description.get("subject", ds_idx)), 0, 0, {}))
            continue

        sfreq = float(ds.raw.info["sfreq"])
        keep_indices = choose_balanced_nonoverlapping_indices(
            annotations=annotations,
            min_gap_samples=min_gap_samples,
            sfreq=sfreq,
            labels_to_keep=labels_to_keep,
        )
        filtered_annotations = _make_annotations_from_indices(annotations, keep_indices)
        ds.raw.set_annotations(filtered_annotations)

        descriptions_after = np.asarray(filtered_annotations.description).astype(str)
        counts_after = {str(label): int(np.sum(descriptions_after == str(label))) for label in labels_to_keep}
        n_after = len(filtered_annotations)
        total_after += n_after
        per_recording.append((ds_idx, str(ds.description.get("subject", ds_idx)), n_before, n_after, counts_after))

    print("CSP overlap filtering:")
    print(f"  Target window samples: {target_window_samples}")
    print(f"  Required onset gap:    {min_gap_samples}")
    print(f"  Events before:         {total_before}")
    print(f"  Events after:          {total_after}")
    print(f"  Events dropped:        {total_before - total_after}")

    for ds_idx, subject_id, n_before, n_after, counts_after in per_recording[:10]:
        print(
            f"    recording={ds_idx}, subject={subject_id}: "
            f"{n_before} -> {n_after}, class_counts={counts_after}"
        )
    if len(per_recording) > 10:
        print(f"    ... {len(per_recording) - 10} more recordings omitted from preview")

    return dataset


def drop_recordings_with_insufficient_class_counts(dataset, labels_to_keep, min_events_per_class):
    """
    Drop recordings that cannot support the configured stratified CV after non-overlap filtering.
    """
    kept_datasets = []
    dropped = []
    labels = [str(label) for label in labels_to_keep]

    for ds_idx, ds in enumerate(dataset.datasets):
        descriptions = np.asarray(ds.raw.annotations.description).astype(str)
        counts = {label: int(np.sum(descriptions == label)) for label in labels}
        if all(counts[label] >= int(min_events_per_class) for label in labels):
            kept_datasets.append(ds)
        else:
            dropped.append((ds_idx, str(ds.description.get("subject", ds_idx)), counts))

    if dropped:
        print("Dropped recordings with too few non-overlapping events for stratified CV:")
        print(f"  Minimum required per class: {min_events_per_class}")
        for ds_idx, subject_id, counts in dropped[:20]:
            print(f"    recording={ds_idx}, subject={subject_id}, class_counts={counts}")
        if len(dropped) > 20:
            print(f"    ... {len(dropped) - 20} more recordings omitted from preview")

    if len(kept_datasets) == 0:
        raise RuntimeError(
            "All recordings were dropped after non-overlap filtering. "
            "Reduce CONFIG['min_events_per_class_after_nonoverlap'], disable the non-overlap filter, "
            "or use a different evaluation design."
        )

    return BaseConcatDataset(kept_datasets)


def compute_trial_stop_offset_samples(filtered_dataset, target_window_samples):
    """
    Compute the stop offset needed to make Braindecode create exactly one fixed-size window per event.

    We intentionally do NOT rewrite MNE annotation durations. For Liu2024, annotations are nominally
    4.0s, but floating-point/sample conversion can produce a shortest annotation of 511 samples.
    Since we intentionally want 537 samples, compute the offset from the shortest selected annotation.
    """
    sfreq = float(filtered_dataset.datasets[0].raw.info["sfreq"])
    timing = print_annotation_timing_diagnostic(
        filtered_dataset,
        target_window_samples=target_window_samples,
        sfreq=sfreq,
    )

    if timing is None:
        raise RuntimeError("No selected annotations remain after filtering.")

    rows = collect_selected_annotation_timing(filtered_dataset)
    duration_samples = np.asarray(
        [int(math.floor(float(r["duration_s"]) * sfreq)) for r in rows],
        dtype=int,
    )

    min_annotation_samples = int(np.min(duration_samples))
    median_annotation_samples = int(np.median(duration_samples))
    max_annotation_samples = int(np.max(duration_samples))
    trial_stop_offset_samples = int(target_window_samples - min_annotation_samples)

    print("Windowing offset configuration:")
    print(f"  Min annotation samples:      {min_annotation_samples}")
    print(f"  Median annotation samples:   {median_annotation_samples}")
    print(f"  Max annotation samples:      {max_annotation_samples}")
    print(f"  Target window samples:       {target_window_samples}")
    print(f"  trial_start_offset_samples:  0")
    print(f"  trial_stop_offset_samples:   {trial_stop_offset_samples}")
    print(f"  on_overlapping_events:       {CONFIG.get('on_overlapping_events', 'raise')}")

    return trial_stop_offset_samples


def build_windows_dataset(dataset, target_duration_s, mapping):
    before_counts = summarize_annotation_counts(dataset)
    print(f"Annotation counts before filtering: {before_counts}")

    total_before, total_after = filter_annotations_by_description(
        dataset,
        descriptions_to_keep=list(mapping.keys()),
    )
    print(f"Filtered annotations: kept {total_after} / {total_before}")

    after_counts = summarize_annotation_counts(dataset)
    print(f"Annotation counts after label filtering: {after_counts}")

    datasets_with_events = [ds for ds in dataset.datasets if len(ds.raw.annotations) > 0]
    dropped_recordings = len(dataset.datasets) - len(datasets_with_events)
    if dropped_recordings > 0:
        print(f"Dropped recordings with zero kept events before non-overlap filtering: {dropped_recordings}")

    if len(datasets_with_events) == 0:
        raise RuntimeError(
            "No recordings contain the selected labels after filtering. "
            "Check CONFIG['labels_to_keep'] and the loaded dataset labels."
        )

    filtered_dataset = BaseConcatDataset(datasets_with_events)
    sfreq = float(filtered_dataset.datasets[0].raw.info["sfreq"])
    window_size_samples = int(math.floor(float(target_duration_s) * sfreq))

    print("\nBefore CSP overlap filtering:")
    _ = print_annotation_timing_diagnostic(
        filtered_dataset,
        target_window_samples=window_size_samples,
        sfreq=sfreq,
    )

    required_nonoverlap_gap_samples = compute_required_nonoverlap_gap_samples(
        filtered_dataset,
        target_window_samples=window_size_samples,
    )

    filtered_dataset = drop_overlapping_annotations_for_csp(
        filtered_dataset,
        target_window_samples=window_size_samples,
        min_gap_samples=required_nonoverlap_gap_samples,
        labels_to_keep=list(mapping.keys()),
    )

    min_events_per_class = int(CONFIG.get("min_events_per_class_after_nonoverlap", CONFIG["cv_folds"]))
    filtered_dataset = drop_recordings_with_insufficient_class_counts(
        filtered_dataset,
        labels_to_keep=list(mapping.keys()),
        min_events_per_class=min_events_per_class,
    )

    print("\nAfter CSP overlap filtering:")
    after_nonoverlap_counts = summarize_annotation_counts(filtered_dataset)
    print(f"Annotation counts after CSP overlap filtering: {after_nonoverlap_counts}")

    trial_stop_offset_samples = compute_trial_stop_offset_samples(
        filtered_dataset,
        target_window_samples=window_size_samples,
    )

    windows_dataset = create_windows_from_events(
        filtered_dataset,
        trial_start_offset_samples=0,
        trial_stop_offset_samples=trial_stop_offset_samples,
        mapping=mapping,
        window_size_samples=window_size_samples,
        window_stride_samples=window_size_samples,
        drop_last_window=True,
        preload=True,
        on_overlapping_events=CONFIG.get("on_overlapping_events", "raise"),
    )

    return windows_dataset
```

## 3.5. Get Subject Windows

```python
def get_subject_windows(windows_dataset, subjects):
    target_subjects = {str(s) for s in subjects}
    subject_to_indices = {str(s): [] for s in subjects}

    for idx, ds in enumerate(windows_dataset.datasets):
        subject_id = str(ds.description.get("subject"))
        if subject_id in target_subjects:
            subject_to_indices[subject_id].append(idx)

    missing_subjects = [sid for sid, idxs in subject_to_indices.items() if len(idxs) == 0]
    if missing_subjects:
        print(f"Subjects with no windows after filtering: {missing_subjects}")

    subject_windows = {}
    for sid in sorted(subject_to_indices.keys(), key=int):
        idxs = subject_to_indices[sid]
        if len(idxs) == 0:
            continue
        subject_windows[sid] = BaseConcatDataset([windows_dataset.datasets[i] for i in idxs])

    if len(subject_windows) == 0:
        raise RuntimeError("No subject windows were created. Ensure windows carry subject metadata and labels exist.")

    return subject_windows

SUBJECT_WINDOWS = get_subject_windows(WINDOWS_DATASET, SUBJECTS)
```

```python
def summarize_subject_windows(subject_windows, n_classes):
    print("Summarizing per-subject windows...")
    for subject_id in sorted(subject_windows, key=int):
        ds = subject_windows[subject_id]
        y = np.asarray([int(ds[i][1]) for i in range(len(ds))], dtype=np.int64)
        counts = np.bincount(y, minlength=n_classes)
        x = ds[0][0]
        print(
            f"  Subject {subject_id}: n_windows={len(ds)}, "
            f"window_shape={tuple(x.shape)}, class_counts={counts.tolist()}"
        )

summarize_subject_windows(SUBJECT_WINDOWS, TARGET_N_CLASSES)
```

# 4. CSP/FBCSP Utilities

## 4.1. Extract Arrays

```python
def get_targets(dataset):
    return np.asarray([int(dataset[i][1]) for i in range(len(dataset))], dtype=np.int64)


def dataset_to_X_y(dataset):
    X = []
    y = []
    for i in range(len(dataset)):
        sample = dataset[i]
        X.append(np.asarray(sample[0], dtype=np.float64))
        y.append(int(sample[1]))
    return np.stack(X, axis=0), np.asarray(y, dtype=np.int64)


def make_fold_splits(y, n_folds, n_classes):
    y = np.asarray(y, dtype=np.int64)
    counts = np.bincount(y, minlength=n_classes)
    min_class_count = int(np.min(counts))

    if min_class_count < n_folds:
        raise ValueError(
            f"Cannot run {n_folds}-fold stratified CV because the smallest class has "
            f"only {min_class_count} samples. Class counts: {counts.tolist()}"
        )

    splitter = StratifiedKFold(
        n_splits=n_folds,
        shuffle=True,
        random_state=BASE_SEED,
    )

    folds = []
    for fold_id, (train_idx, test_idx) in enumerate(splitter.split(np.zeros(len(y)), y), start=1):
        folds.append({"fold_id": fold_id, "idx_train": train_idx, "idx_test": test_idx})
    return folds
```

## 4.2. Filtering and Classifiers

```python
def bandpass_epochs(X, l_freq, h_freq, sfreq):
    """Filter epoched data with shape (n_trials, n_channels, n_times)."""
    return mne.filter.filter_data(
        X.copy(),
        sfreq=float(sfreq),
        l_freq=float(l_freq) if l_freq is not None else None,
        h_freq=float(h_freq) if h_freq is not None else None,
        method=CONFIG["feature_filter_method"],
        iir_params=dict(CONFIG["feature_iir_params"]),
        n_jobs=1,
        verbose=False,
    )


def make_classifier():
    classifier_name = CONFIG["classifier"].lower()
    if classifier_name == "lda":
        # shrinkage LDA is more stable for small within-subject MI folds
        return LinearDiscriminantAnalysis(solver="lsqr", shrinkage="auto")
    if classifier_name == "linear_svm":
        return SVC(
            kernel="linear",
            C=float(CONFIG["linear_svm_C"]),
            class_weight="balanced",
            probability=True,
            random_state=BASE_SEED,
        )
    raise ValueError(f"Unknown classifier: {CONFIG['classifier']}")


def get_score_values(estimator, X_features_or_raw):
    """Return a continuous class-1 score for ROC-AUC when available."""
    try:
        if hasattr(estimator, "decision_function"):
            scores = estimator.decision_function(X_features_or_raw)
            scores = np.asarray(scores)
            if scores.ndim == 2 and scores.shape[1] > 1:
                return scores[:, 1]
            return scores.reshape(-1)
        if hasattr(estimator, "predict_proba"):
            probs = estimator.predict_proba(X_features_or_raw)
            return np.asarray(probs)[:, 1]
    except Exception:
        return None
    return None


def safe_roc_auc(y_true, score_values):
    if score_values is None:
        return None
    y_true = np.asarray(y_true)
    if len(np.unique(y_true)) < 2:
        return None
    try:
        return float(roc_auc_score(y_true, score_values))
    except Exception:
        return None
```

## 4.3. CSP and FBCSP Fit/Predict

```python
def make_csp_transformer(n_components):
    return CSP(
        n_components=int(n_components),
        reg=CONFIG["csp"]["reg"],
        log=bool(CONFIG["csp"]["log"]),
        norm_trace=bool(CONFIG["csp"]["norm_trace"]),
    )


def fit_predict_csp(X_train, y_train, X_test):
    csp_cfg = CONFIG["csp"]
    if csp_cfg["apply_feature_filter"]:
        X_train_use = bandpass_epochs(
            X_train,
            l_freq=csp_cfg["l_freq"],
            h_freq=csp_cfg["h_freq"],
            sfreq=CONFIG["sfreq"],
        )
        X_test_use = bandpass_epochs(
            X_test,
            l_freq=csp_cfg["l_freq"],
            h_freq=csp_cfg["h_freq"],
            sfreq=CONFIG["sfreq"],
        )
        band = [float(csp_cfg["l_freq"]), float(csp_cfg["h_freq"])]
    else:
        X_train_use = X_train
        X_test_use = X_test
        band = [float(CONFIG["bandpass_low"]), float(CONFIG["bandpass_high"])]

    n_components = min(int(csp_cfg["n_components"]), X_train_use.shape[1])

    estimator = Pipeline(
        steps=[
            ("csp", make_csp_transformer(n_components)),
            ("scaler", StandardScaler()),
            ("classifier", make_classifier()),
        ]
    )

    estimator.fit(X_train_use, y_train)
    y_pred = estimator.predict(X_test_use)
    score_values = get_score_values(estimator, X_test_use)

    extra = {
        "feature_band": band,
        "n_csp_components": n_components,
        "n_features": n_components,
    }
    return y_pred, score_values, extra


def fit_predict_fbcsp(X_train, y_train, X_test):
    fb_cfg = CONFIG["fbcsp"]
    train_features = []
    test_features = []
    used_bands = []

    for band in fb_cfg["bands"]:
        l_freq, h_freq = float(band[0]), float(band[1])
        X_train_band = bandpass_epochs(X_train, l_freq=l_freq, h_freq=h_freq, sfreq=CONFIG["sfreq"])
        X_test_band = bandpass_epochs(X_test, l_freq=l_freq, h_freq=h_freq, sfreq=CONFIG["sfreq"])

        n_components = min(int(fb_cfg["n_components"]), X_train_band.shape[1])
        csp = CSP(
            n_components=n_components,
            reg=fb_cfg["reg"],
            log=bool(fb_cfg["log"]),
            norm_trace=bool(fb_cfg["norm_trace"]),
        )

        train_features.append(csp.fit_transform(X_train_band, y_train))
        test_features.append(csp.transform(X_test_band))
        used_bands.append([l_freq, h_freq])

    if len(train_features) == 0:
        raise RuntimeError("No FBCSP bands produced features.")

    X_train_features = np.concatenate(train_features, axis=1)
    X_test_features = np.concatenate(test_features, axis=1)

    estimator = Pipeline(
        steps=[
            ("scaler", StandardScaler()),
            ("classifier", make_classifier()),
        ]
    )
    estimator.fit(X_train_features, y_train)
    y_pred = estimator.predict(X_test_features)
    score_values = get_score_values(estimator, X_test_features)

    extra = {
        "feature_bands": used_bands,
        "n_bands": len(used_bands),
        "n_csp_components_per_band": int(fb_cfg["n_components"]),
        "n_features": int(X_train_features.shape[1]),
    }
    return y_pred, score_values, extra
```

# 5. Cross-Validation

## 5.1. Single Fold Runner

```python
def run_single_fold(model_name, subject_id, X, y, idx_train, idx_test, fold_id):
    X_train = X[idx_train]
    X_test = X[idx_test]
    y_train = y[idx_train]
    y_test = y[idx_test]

    train_counts = np.bincount(y_train, minlength=TARGET_N_CLASSES)
    test_counts = np.bincount(y_test, minlength=TARGET_N_CLASSES)

    if PRINT_FOLD_CLASS_COUNTS:
        print(
            f"    Fold {fold_id}: train={len(idx_train)} {train_counts.tolist()} | "
            f"test={len(idx_test)} {test_counts.tolist()}"
        )

    if model_name == "CSP_LDA":
        y_pred, score_values, extra = fit_predict_csp(X_train, y_train, X_test)
    elif model_name == "FBCSP_LDA":
        y_pred, score_values, extra = fit_predict_fbcsp(X_train, y_train, X_test)
    else:
        raise ValueError(f"Unknown model_name: {model_name}")

    acc = float(accuracy_score(y_test, y_pred))
    bal_acc = float(balanced_accuracy_score(y_test, y_pred))
    roc_auc = safe_roc_auc(y_test, score_values)
    cm = confusion_matrix(y_test, y_pred, labels=list(range(TARGET_N_CLASSES))).tolist()
    pred_hist = np.bincount(np.asarray(y_pred, dtype=np.int64), minlength=TARGET_N_CLASSES).tolist()

    return {
        "subject_id": str(subject_id),
        "fold_id": int(fold_id),
        "paradigm": CONFIG["paradigm"],
        "model_name": model_name,
        "classifier": CONFIG["classifier"],
        "n_train": int(len(idx_train)),
        "n_test": int(len(idx_test)),
        "train_class_counts": train_counts.tolist(),
        "test_class_counts": test_counts.tolist(),
        "accuracy": acc,
        "balanced_accuracy": bal_acc,
        "roc_auc": roc_auc,
        "confusion_matrix": cm,
        "prediction_histogram": pred_hist,
        **extra,
    }
```

## 5.2. Subject CV Runner

```python
def run_subject_cv(model_name, subject_id, subject_dataset, n_classes, cv_folds):
    X, y = dataset_to_X_y(subject_dataset)
    counts = np.bincount(y, minlength=n_classes)
    print(f"\n[{model_name}] Subject {subject_id}: {len(subject_dataset)} windows | class_counts={counts.tolist()}")

    folds = make_fold_splits(y, n_folds=cv_folds, n_classes=n_classes)
    results = []

    for fold in folds:
        result = run_single_fold(
            model_name=model_name,
            subject_id=subject_id,
            X=X,
            y=y,
            idx_train=fold["idx_train"],
            idx_test=fold["idx_test"],
            fold_id=fold["fold_id"],
        )
        results.append(result)

    acc_values = [r["accuracy"] for r in results if r["accuracy"] is not None]
    bal_acc_values = [r["balanced_accuracy"] for r in results if r["balanced_accuracy"] is not None]
    roc_auc_values = [r["roc_auc"] for r in results if r["roc_auc"] is not None]

    mean_acc = float(np.mean(acc_values)) if len(acc_values) > 0 else None
    std_acc = float(np.std(acc_values)) if len(acc_values) > 0 else None
    mean_bal_acc = float(np.mean(bal_acc_values)) if len(bal_acc_values) > 0 else None
    std_bal_acc = float(np.std(bal_acc_values)) if len(bal_acc_values) > 0 else None
    mean_roc_auc = float(np.mean(roc_auc_values)) if len(roc_auc_values) > 0 else None
    std_roc_auc = float(np.std(roc_auc_values)) if len(roc_auc_values) > 0 else None

    acc_str = f"{mean_acc:.4f}±{std_acc:.4f}" if mean_acc is not None and std_acc is not None else "N/A"
    bal_acc_str = f"{mean_bal_acc:.4f}±{std_bal_acc:.4f}" if mean_bal_acc is not None and std_bal_acc is not None else "N/A"
    roc_auc_str = f"{mean_roc_auc:.4f}±{std_roc_auc:.4f}" if mean_roc_auc is not None and std_roc_auc is not None else "N/A"

    print(f"  Subject {subject_id} summary: acc={acc_str}  bal_acc={bal_acc_str}  roc_auc={roc_auc_str}")
    return results
```

## 5.3. Run All Subjects

```python
print("=" * 70)
print("STARTING CSP/FBCSP CROSS-VALIDATION")
print("=" * 70)
print(f"Models: {CONFIG['run_models']}")
print(f"Subjects: {len(SUBJECT_WINDOWS)}")
print(f"CV folds: {CONFIG['cv_folds']}")

ALL_MODEL_RESULTS = {}

for model_name in CONFIG["run_models"]:
    print("\n" + "=" * 70)
    print(f"RUNNING MODEL: {model_name}")
    print("=" * 70)

    fold_results = []
    for sid in sorted(SUBJECT_WINDOWS.keys(), key=int):
        subject_results = run_subject_cv(
            model_name=model_name,
            subject_id=sid,
            subject_dataset=SUBJECT_WINDOWS[sid],
            n_classes=TARGET_N_CLASSES,
            cv_folds=CONFIG["cv_folds"],
        )
        fold_results.extend(subject_results)

    ALL_MODEL_RESULTS[model_name] = fold_results
    print(f"\nCompleted {model_name}: {len(fold_results)} folds")

print("\nAll requested CSP/FBCSP models completed.")
```

# 6. Results

## 6.1. Aggregate Metrics

```python
def aggregate_fold_results(fold_results):
    subject_metrics = {}

    for result in fold_results:
        sid = result["subject_id"]
        if sid not in subject_metrics:
            subject_metrics[sid] = {
                "accuracies": [],
                "balanced_accuracies": [],
                "roc_aucs": [],
            }
        subject_metrics[sid]["accuracies"].append(result["accuracy"])
        subject_metrics[sid]["balanced_accuracies"].append(result["balanced_accuracy"])
        subject_metrics[sid]["roc_aucs"].append(result.get("roc_auc"))

    for sid, m in subject_metrics.items():
        acc_values = [v for v in m["accuracies"] if v is not None]
        bal_acc_values = [v for v in m["balanced_accuracies"] if v is not None]
        roc_auc_values = [v for v in m["roc_aucs"] if v is not None]

        m["mean_accuracy"] = float(np.mean(acc_values)) if len(acc_values) > 0 else None
        m["std_accuracy"] = float(np.std(acc_values)) if len(acc_values) > 0 else None
        m["mean_balanced_accuracy"] = float(np.mean(bal_acc_values)) if len(bal_acc_values) > 0 else None
        m["std_balanced_accuracy"] = float(np.std(bal_acc_values)) if len(bal_acc_values) > 0 else None
        m["mean_roc_auc"] = float(np.mean(roc_auc_values)) if len(roc_auc_values) > 0 else None
        m["std_roc_auc"] = float(np.std(roc_auc_values)) if len(roc_auc_values) > 0 else None

    all_accs = [r["accuracy"] for r in fold_results if r["accuracy"] is not None]
    all_bal_accs = [r["balanced_accuracy"] for r in fold_results if r["balanced_accuracy"] is not None]
    all_roc_aucs = [r["roc_auc"] for r in fold_results if r.get("roc_auc") is not None]

    global_metrics = {
        "mean_accuracy": float(np.mean(all_accs)) if len(all_accs) > 0 else None,
        "std_accuracy": float(np.std(all_accs)) if len(all_accs) > 0 else None,
        "mean_balanced_accuracy": float(np.mean(all_bal_accs)) if len(all_bal_accs) > 0 else None,
        "std_balanced_accuracy": float(np.std(all_bal_accs)) if len(all_bal_accs) > 0 else None,
        "mean_roc_auc": float(np.mean(all_roc_aucs)) if len(all_roc_aucs) > 0 else None,
        "std_roc_auc": float(np.std(all_roc_aucs)) if len(all_roc_aucs) > 0 else None,
        "n_subjects": len(subject_metrics),
        "n_folds_total": len(fold_results),
    }

    return subject_metrics, global_metrics

SUBJECT_METRICS = {}
GLOBAL_METRICS = {}

for model_name, fold_results in ALL_MODEL_RESULTS.items():
    subject_metrics, global_metrics = aggregate_fold_results(fold_results)
    SUBJECT_METRICS[model_name] = subject_metrics
    GLOBAL_METRICS[model_name] = global_metrics

print("=" * 70)
print("AGGREGATED RESULTS")
print("=" * 70)

for model_name in CONFIG["run_models"]:
    gm = GLOBAL_METRICS[model_name]
    acc_str = f"{gm['mean_accuracy']:.4f}±{gm['std_accuracy']:.4f}" if gm["mean_accuracy"] is not None and gm["std_accuracy"] is not None else "N/A"
    bal_acc_str = f"{gm['mean_balanced_accuracy']:.4f}±{gm['std_balanced_accuracy']:.4f}" if gm["mean_balanced_accuracy"] is not None and gm["std_balanced_accuracy"] is not None else "N/A"
    roc_auc_str = f"{gm['mean_roc_auc']:.4f}±{gm['std_roc_auc']:.4f}" if gm["mean_roc_auc"] is not None and gm["std_roc_auc"] is not None else "N/A"
    print(f"  {model_name}: acc={acc_str}  bal_acc={bal_acc_str}  roc_auc={roc_auc_str}  folds={gm['n_folds_total']}")

print("=" * 70)
```

## 6.2. Prediction Collapse Diagnostics

```python
def summarize_prediction_collapse(fold_results):
    total = len(fold_results)
    collapsed = 0
    majority_only = {str(c): 0 for c in range(TARGET_N_CLASSES)}

    for r in fold_results:
        hist = r.get("prediction_histogram", [])
        nonzero = sum(1 for v in hist if v > 0)
        if nonzero <= 1:
            collapsed += 1
            if len(hist) > 0:
                majority_only[str(int(np.argmax(hist)))] += 1

    return {
        "n_folds": total,
        "n_collapsed_single_class": collapsed,
        "collapsed_fraction": float(collapsed / total) if total > 0 else None,
        "single_class_prediction_counts": majority_only,
    }

COLLAPSE_DIAGNOSTICS = {}
for model_name, fold_results in ALL_MODEL_RESULTS.items():
    COLLAPSE_DIAGNOSTICS[model_name] = summarize_prediction_collapse(fold_results)

print("Prediction collapse diagnostics")
print(json.dumps(COLLAPSE_DIAGNOSTICS, indent=2))
```
